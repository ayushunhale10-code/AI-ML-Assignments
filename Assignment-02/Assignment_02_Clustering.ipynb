# Assignment 2: Clustering Using Unsupervised Learning

## Aim

To apply clustering techniques using unsupervised learning and group similar data points into clusters.

## Theory

Clustering is an unsupervised machine learning technique used to group similar data points into clusters.

In unsupervised learning, the data does not have predefined output labels. The algorithm identifies patterns and groups the data based on similarity.

K-Means is a popular clustering algorithm. It divides the data into a specified number of clusters by assigning each data point to the nearest cluster center (centroid).

In this assignment, K-Means clustering is used to divide the data points into different groups and visualize the resulting clusters.

```python
# Import required libraries

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
# Create a sample dataset

data = {
    "Age": [18, 19, 20, 21, 22, 35, 36, 37, 38, 39, 50, 51, 52, 53, 54],
    "Income": [20, 22, 24, 25, 23, 45, 47, 50, 48, 46, 70, 72, 75, 73, 76]
}

df = pd.DataFrame(data)

print("Dataset:")
display(df)
```

```python
# Apply K-Means clustering

X = df[["Age", "Income"]]

kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)

df["Cluster"] = kmeans.fit_predict(X)

print("Clustered Dataset:")
display(df)
```

```python
# Visualize the clusters

plt.figure(figsize=(8, 5))

plt.scatter(
    df["Age"],
    df["Income"],
    c=df["Cluster"],
    s=100
)

plt.xlabel("Age")
plt.ylabel("Income")
plt.title("K-Means Clustering")
plt.grid(True)

plt.show()
```

## Result Interpretation

The K-Means clustering algorithm successfully grouped the data points into three clusters based on Age and Income.

The data points with similar Age and Income values were placed into the same cluster. The scatter plot provides a visual representation of the different clusters.

Thus, K-Means clustering can be used to identify groups or patterns in unlabeled data.

## Result

K-Means clustering was successfully applied to the dataset.

The data points were divided into three clusters based on Age and Income. The clusters were visualized using a scatter plot, which clearly shows the grouping of similar data points.

Hence, unsupervised learning using K-Means clustering was successfully implemented.

## Conclusion

K-Means is a useful unsupervised learning algorithm for grouping similar data points into clusters. In this experiment, the data was successfully divided into three clusters based on Age and Income.

The clustering results demonstrate how unsupervised learning can discover patterns and groups in data without predefined labels.
