# Assignment 3: Dimensionality Reduction Using PCA

## Aim

To apply Principal Component Analysis (PCA) for dimensionality reduction and visualize high-dimensional data in a lower-dimensional space.

## Theory

Principal Component Analysis (PCA) is a dimensionality reduction technique used to reduce the number of features in a dataset while preserving as much important information as possible.

PCA transforms the original features into a new set of variables called principal components. The first principal component contains the maximum possible variance, followed by the second component, and so on.

In this assignment, PCA is applied to the Iris dataset to reduce its four features into two principal components and visualize the data in a two-dimensional space.

```python
# Import required libraries

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
# Load the Iris dataset

iris = load_iris()

X = iris.data
y = iris.target

print("Dataset shape:", X.shape)
print("Feature names:", iris.feature_names)
```

```python
# Standardize the features

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

print("Standardization completed successfully.")
```

```python
# Apply PCA and reduce the data to 2 dimensions

pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

print("PCA completed successfully.")
print("Reduced dataset shape:", X_pca.shape)
```

```python
# Visualize the PCA result

plt.figure(figsize=(8, 5))

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=y,
    s=60
)

plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.title("PCA of Iris Dataset")
plt.grid(True)

plt.show()
```

## Result Interpretation

PCA successfully reduced the original four features of the Iris dataset to two principal components.

The scatter plot shows that the data points form separate groups in the reduced two-dimensional space. This demonstrates that PCA can reduce the number of dimensions while retaining important patterns in the data.

## Result

PCA was successfully applied to the Iris dataset.

The original four features were reduced to two principal components, making the dataset easier to visualize and analyze. The PCA plot shows the distribution of the Iris data in a lower-dimensional space.

Hence, dimensionality reduction using PCA was successfully implemented.

## Conclusion

PCA is an effective dimensionality reduction technique that transforms a large number of features into a smaller number of principal components.

In this experiment, the four features of the Iris dataset were reduced to two principal components while preserving important patterns in the data.
